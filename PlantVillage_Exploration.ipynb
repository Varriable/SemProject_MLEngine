## Expected Output Variables

1. **PlantVillage:** The expected output is the **plant disease/condition label**, such as `Tomato___Early_blight` or `Apple___healthy`.

2. **Plant Leaves for Image Classification:** The expected output is the **health status of the leaf**, classified as either **healthy** or **diseased**.

3. **Dataset for Crop Pest and Disease Detection:** The expected output is the **pest or disease class**, such as `anthracnose`, `bacterial blight`, `leaf blight`, or `healthy`.

4. **Plant Disease Detection:** The expected output is the **predicted plant disease or healthy condition** of the leaf image.

```python
#This is to initialise the pandas and the dataset

import pandas as pd
from datasets import load_dataset

ds = load_dataset("mohanty/PlantVillage","default")
```

```python
#Here we separate the 2 groups found in the dataset.

train = ds["train"]
test = ds["test"]
```

```python
#Here we find the number of columns and number of rows found in the dataset

columns = train.num_columns
rows = train.num_rows
print(f'The number of coluns are: {columns} while the number of rows are {rows}')
```

```python
#Here we find the datatypes found in the dataset

datatypes = train.features
print(datatypes)
```

```python
#Here we convert to a pandas dataframe and check for null values
df = pd.DataFrame(train)
null_count = df.isnull().sum()
print(null_count)
```

```python
#Here we combine the first 15 rows and last 20 rows and display that dataframe

first_15 = train[:15]
last_20 = train[-20:]
data = {**first_15 , **last_20}
df_sample =pd.DataFrame(data)
print(df_sample)
```
